# what is KNN?

K-NN is a non-parametric and lazy learning algorithm. Non-Prametric means there is no assumption for underlying data distribution that is the model structure is detrmined from the dataset.

It is called Lazy alg because it does not need any training data points for model generation.all training data is used in the testing phase which makes taining faster and testing phase slower and costlier.

K-Nearest Neighbour (K-NN) is a sample alg that stored all the available casea and classifies  the new data or case based on a similarity measure.

### What is K-Nearest Neighbors (KNN)?
K-Nearest Neighbors (KNN) is a supervised machine learning algorithm used for classification and regression tasks. It is a lazy learning and instance-based algorithm, meaning it doesn't learn from the data during training but memorizes it and makes decisions during prediction.

### How Does KNN Work?

Choose the number of neighbors k.

Calculate the distance between the new data point and all points in the training dataset.

Common distance metrics:
Euclidean Distance (most common)
Manhattan Distance
Minkowski Distance
Select the k nearest neighbors (the points closest to the new data point).
Make Predictions:

For Classification:

Assign the class most common among the k neighbors (majority vote).

For Regression:

Calculate the average of the values of the k neighbors.
    
### KNN Classification Example

Scenario: Classifying whether a fruit is an apple or orange based on features like weight and color.
Steps:

Calculate the distance between the new fruit and all fruits in the dataset.
    
Choose k nearest fruits.
    
Count the most frequent class among these k fruits.
    
Classify the new fruit based on the majority class.

# why its lazy algorithm and why does not learn from the data.

### No Training Phase:-------->

Unlike other algorithms (like Decision Trees or Logistic Regression) that learn patterns and build a model during training, KNN does not perform any learning or model-building when it's given the training data.
Instead, it simply stores the entire dataset and waits until a prediction is required.

### Learning Happens at Prediction Time:---------->

When a new data point is introduced, KNN calculates the distance between this point and all the points in the stored dataset.
It then selects the 'k' closest neighbors and makes a prediction based on their values.
This is why it's called lazy—it "delays" the learning process until it absolutely needs to make a prediction.

### Why Doesn't KNN Learn During Training?---------->
 
Instance-Based Learning: KNN is an instance-based algorithm, meaning it relies directly on the data instances for prediction without forming an abstract model.

No Assumptions: KNN doesn't assume any underlying data distribution or relationship between features and output.

Storage Approach: It simply stores the data as it is and uses it during prediction by comparing distances.

### What is the Optimal Value of k in KNN?

In K-Nearest Neighbors (KNN), the value of k (number of neighbors) is a crucial hyperparameter that directly affects the model's performance.

A small k (like 1 or 3) makes the model sensitive to noise and can lead to overfitting.
A large k smooths out predictions but can lead to underfitting, missing important patterns in the data.

### How to Choose the Optimal k?

Start with an Odd Value:
Use an odd number to avoid ties in binary classification problems.
    
Use the Square Root Rule (as a starting point):
k=square root of (n)
Where 
n is the total number of data points.
Example: If you have 100 samples, start with 
𝑘=10

Use Cross-Validation:

Perform k-fold cross-validation to test different k values and find the one with the lowest error rate.

Elbow Method (for Classification):

Plot the error rate vs. different k values.
The optimal k is at the "elbow point" where the error rate starts to flatten.

Check Accuracy or Error Rate:

Choose k that provides the highest accuracy or lowest error on validation data.

```python
df=pd.DataFrame(np.c_[cancer['data'], cancer['target']], columns=np.append(cancer['feature_names'],['target']))
```

```python
x=df.iloc[:,:-1].values
y=df.iloc[:,-1].values
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test=train_test_split(x, y, test_size=0.2, random_state=0)
```

```python
x_train.shape
```

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
x_train=sc.fit_transform(x_train)
x_test=sc.transform(x_test)
```

```python
x_train.shape
```

```python
from sklearn.neighbors import KNeighborsClassifier
cls=KNeighborsClassifier(n_neighbors=22)
```

```python
cls.fit(x_train, y_train)
```

```python
y_pred=cls.predict(x_test)
y_pred
```

```python
from sklearn.metrics import confusion_matrix,accuracy_score
cm=confusion_matrix(y_test, y_pred)
sns.heatmap(cm, annot=True)
```

```python
accuracy_score(y_test, y_pred)
```
