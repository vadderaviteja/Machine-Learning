# Difference between Homogeneous and Heterogeneous Ensemble


# Homogeneous Ensemble Learning

1. Homogeneous Ensemble Learning

Homogeneous ensemble learning uses multiple models of the same algorithm type but trained differently (e.g., on different subsets of data).

Example Algorithms:
Random Forest (many decision trees)

Bagging with Decision Trees

How it Works:
Same base learner (e.g., Decision Tree) is trained multiple times with variations (e.g., different data samples).

Outputs are combined (like majority voting or averaging).

Purpose:
Reduce variance (overfitting) and improve model stability.

Analogy:
Like asking the same expert multiple times, each with slightly different information.




# Heterogeneous Ensemble Learning

🔹 2. Heterogeneous Ensemble Learning ✅ Definition: Heterogeneous ensemble learning uses different types of algorithms (e.g., SVM, Decision Tree, Logistic Regression) combined together.

Example Algorithms: Stacking (Stacked Generalization)

VotingClassifier with mixed models

How it Works: Diverse models are trained independently.

Their outputs are combined, usually by a meta-model (in stacking) or voting (in classification).

Purpose: Leverage strengths of different algorithms to increase overall accuracy.

Analogy: Like consulting a group of experts from different fields — each brings unique value to the decision.
