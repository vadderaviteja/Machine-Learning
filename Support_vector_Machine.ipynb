# Support vector Machine

A Support Vector Machine (SVM) is a supervised learning algorithm used for both classification and regression tasks. It works by finding the optimal hyperplane in an N-dimensional space (where N is the number of features) that best separates the data into different classes.

In simpler terms, SVM tries to draw a boundary (line, plane, or surface) that clearly divides the data into different categories, while maximizing the distance (margin) between that boundary and the nearest data points of each class, which are called support vectors.

# HYPERPLANE

A hyperplane is a decision boundary that helps separate data points into different classes in machine learning models, especially in algorithms like Support Vector Machines (SVM).

# Implementation of SVM REgression

```python

import numpy as np
import pandas as pd

     

df=pd.read_csv(r"C:\Users\Windows\Downloads\Data.csv")
     
```

```python
df.isna().sum()
```

```python
x=df.iloc[:, :-1].values
y=df.iloc[:, -1:].values
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test =train_test_split(x, y, test_size=0.2, random_state=0)
```

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
x_train = sc.fit_transform(x_train)
x_test=sc.transform(x_test)
y_train =sc.fit_transform(y_train)
```

```python
x_train
```

```python
x_test
```

```python
y_train
```

```python
from sklearn.svm import SVR
rg=SVR()
rg.fit(x_train, y_train)
```

```python
y_pred = rg.predict(x_test)
```

```python
y_pred
```

```python
y_pred =sc.inverse_transform(y_pred.reshape(-1,1))
```

```python
y_pred
```

```python
from sklearn.metrics import r2_score
print(r2_score(y_test, y_pred))
```

# Implemetation of SVM Classification

```python
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
from sklearn.datasets import load_iris
df=load_iris()
```

```python
df
```

```python
df.keys()
```

```python
print(df['DESCR'])
```

```python
df1=pd.DataFrame(np.c_[df['data'],df['target']], columns=(np.append(df['feature_names'], ['target'] )) )
```

```python
df1.head()
```

```python
x=df1.iloc[:, :-1].values
y=df1.iloc[:, -1:].values
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test= train_test_split(x, y, test_size=0.2, random_state=0)
```

```python
from sklearn.svm import SVC
cf=SVC(kernel='rbf')
cf.fit(x_train, y_train)
```

```python
y_pred=cf.predict(x_test)
```

```python
y_pred
```

```python
from sklearn.metrics import confusion_matrix, accuracy_score
cm=confusion_matrix(y_test, y_pred)
sns.heatmap(cm, annot=True)
```

```python
accuracy_score(y_pred, y_test)
```
