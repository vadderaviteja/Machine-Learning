```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv(r"Mall_Customers.csv")
```

```python
df.shape
```

```python
df.isna().sum()
```

```python
df.head()
```

```python
X=df.iloc[:, [3, 4]].values
X[:10]
```

```python
import warnings
warnings.filterwarnings('ignore')
from sklearn.cluster import KMeans
WCSS=[]
for i in range(1, 11):
    kk=KMeans(n_clusters=i, init='k-means++', random_state=0)
    kk.fit(X)
    WCSS.append(kk.inertia_)

plt.plot(range(1,11), WCSS)
plt.title('the ELBOW METHOD')
plt.xlabel('no of clusters')
plt.ylabel('WCSS')
plt.show()
    
                  
```

```python

kk=KMeans(n_clusters=i, init='k-means++', random_state=0)
y_kmeans=kk.fit_predict(X)
```

```python
y_kmeans
```

```python
plt.scatter(X[y_kmeans==0, 0], X[y_kmeans==0,1], s=100, c='red', label='cluster1')
plt.scatter(X[y_kmeans==1, 0], X[y_kmeans==1,1], s=100, c='blue', label='cluster1')
plt.scatter(X[y_kmeans==2, 0], X[y_kmeans==2,1], s=100, c='green', label='cluster1')
plt.scatter(X[y_kmeans==3, 0], X[y_kmeans==3,1], s=100, c='yellow', label='cluster1')
plt.scatter(X[y_kmeans==4, 0], X[y_kmeans==4,1], s=100, c='brown', label='cluster1')
plt.xlabel('Anual Income')
plt.ylabel('Spending score')
plt.show()
```
