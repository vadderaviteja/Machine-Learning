# standarscaler method  using multiple features

```python
import warnings
warnings.filterwarnings('ignore')
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

# Sample Data: Height (cm) and Weight (kg)
data = np.array([
    [150, 50],
    [160, 60],
    [170, 70],
    [180, 80],
    [190, 90],
    [200, 100]
])

# Convert to DataFrame for clarity
df = pd.DataFrame(data, columns=["Height", "Weight"])
print("Original Data:\n", df)

# --------------------------
# 1. Without Standardization
# --------------------------
kmeans_original = KMeans(n_clusters=2, random_state=42)
labels_original = kmeans_original.fit_predict(df)

# --------------------------
# 2. With Standardization
# --------------------------
scaler = StandardScaler()
standardized_data = scaler.fit_transform(df)

kmeans_standardized = KMeans(n_clusters=2, random_state=42)
labels_standardized = kmeans_standardized.fit_predict(standardized_data)

# --------------------------
# Plotting the Results
# --------------------------
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

# Plot without standardization
ax1.scatter(df["Height"], df["Weight"], c=labels_original, cmap='viridis', marker='o')
ax1.set_title("Without Standardization")
ax1.set_xlabel("Height (cm)")
ax1.set_ylabel("Weight (kg)")

# Plot with standardization
ax2.scatter(standardized_data[:, 0], standardized_data[:, 1], c=labels_standardized, cmap='viridis', marker='o')
ax2.set_title("With Standardization")
ax2.set_xlabel("Standardized Height")
ax2.set_ylabel("Standardized Weight")

plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv(r"C:\Users\Windows\ML_Theory\Wine.csv")
```

```python
df.head()
```

```python
df.Customer_Segment.unique()
```

```python
df.isna().sum()
```

```python
x=df.iloc[:, :-1].values
y=df.iloc[:, -1].values
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test=train_test_split(x, y, test_size=0.2, random_state=0)
```

```python
from sklearn.linear_model import LogisticRegression
reg=LogisticRegression()
reg.fit(x_train, y_train)
```

```python
y_pred=reg.predict(x_test)
y_pred
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix
cm=confusion_matrix(y_pred, y_test)
cm
```

```python
sns.heatmap(cm, annot=True,)
```

```python
sns.heatmap(cm, annot=True,)
```

# After a Dimensionality Reduction

```python
from sklearn.preprocessing import StandardScaler
sd=StandardScaler()

sd
```

```python
x_train=sd.fit_transform(x_train)
x_test=sd.transform(x_test)
```

```python
from sklearn.decomposition import PCA
pca=PCA(n_components=2)
pca


```

```python
x_train=pca.fit_transform(x_train)
x_test=pca.transform(x_test)
```

```python
variance=pca.explained_variance_ratio_
variance
```

```python
from sklearn.linear_model import LogisticRegression
re=LogisticRegression()
re.fit(x_train, y_train)
```

```python
y_pr=re.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix
cm=confusion_matrix(y_pr, y_test)
cm
```

```python
accuracy_score(y_pr,y_test)
```

```python
from matplotlib.colors import ListedColormap
x_set, y_set=x_train, y_train
x1, x2=np.meshgrid(np.arange(start=x_set[:, 0].min()-1, stop=x_set[:, 0].max() + 1, step = 0.01), 
                   np.arange(start=x_set[:, 1].min()-1, stop=x_set[:, 1].max() + 1, step = 0.01))

plt.contourf(x1, x2, re.predict(np.array([x1.ravel(), x2.ravel()]).T).reshape(x1.shape), alpha=0.75, cmap=ListedColormap(('red', 'green', 'blue')))

plt.xlim(x1.min(), x1.max())
plt.ylim(x2.min(), x2.max())
for i , j in enumerate(np.unique(y_set)):
    plt.scatter(x_set[y_set == j, 0], x_set[y_set == j, 1], 
               c=ListedColormap(('red', 'green', 'blue'))(i), label=j)

plt.title('Logistic Regrssion (Training set)')
plt.xlabel('pc1')
plt.ylabel('pc2')
plt.legend()
plt.show()
```
