### What is the Random Forest Algorithm?

Random Forest is a powerful ensemble machine learning algorithm used for both classification and regression tasks. It works by building multiple decision trees and combining their outputs to improve accuracy and robustness.

It is based on the concept of "wisdom of the crowd" — combining multiple weak models (trees) to make a strong model.

It uses a technique called bagging (Bootstrap Aggregation) to train each tree on random subsets of data.

# STEP  BY STEP PROCESS

### Step 1: Prepare the Dataset

You must have a labeled dataset with:

Input features (X): e.g., size, rooms, location

Target variable (y): e.g., house price (continuous)

### Step 2: Create Bootstrap Samples (Bagging)

From the original dataset, the algorithm randomly selects N samples with replacement to create multiple bootstrap datasets.

Example: From a dataset of 100 rows, it creates 10 random datasets (say, for 10 trees).

 Why with replacement?

Some rows may repeat, and some may be left out (called OOB – Out Of Bag samples, used later for error estimation).

### Step 3: Train Decision Trees

For each bootstrap sample:

Train a regression decision tree.

At each node/split, consider only a random subset of features (not all features).

This creates diverse trees and reduces overfitting.

Example:

If you have 8 features, each tree may use only 3–4 randomly chosen features for splitting.

### Step 4: Make Predictions from All Trees

When you want to predict for a new input:

Each tree gives its own regression output (e.g., 10 different predicted house prices).

### Step 5: Aggregate the Predictions

Combine the output of all trees by taking the average:

* This reduces noise and gives a more stable and accurate prediction.

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.tree import plot_tree
import warnings

warnings.filterwarnings('ignore')
rg=RandomForestRegressor(n_estimators=10, random_state=0)
rg.fit(x_train, y_train)
```
