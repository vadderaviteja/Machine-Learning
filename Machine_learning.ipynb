# labeled Data

```python
from sklearn.linear_model import LinearRegression
import numpy as np

# Labeled data: Square footage (input) and House prices (output)
X = np.array([[1000], [1500], [2000], [2500], [3000]])  # Square footage
y = np.array([200000, 250000, 300000, 350000, 400000])  # House price

# Train a Linear Regression model
model = LinearRegression()
model.fit(X, y)

# Predict the price of a 2200 sq ft house
predicted_price = model.predict([[2200]])
print(f"Predicted price for 2200 sq ft house: ${predicted_price[0]:,.2f}")
```

# Pattern Without Labeled Data (Unsupervised Learning)

```python
from sklearn.cluster import KMeans
import numpy as np

# Unlabeled data: Customer income and spending
X = np.array([[25000, 400], [45000, 700], [50000, 900], [60000, 1200], [80000, 2000]])

# Create a K-Means model with 2 clusters (groups)
kmeans = KMeans(n_clusters=2, random_state=42)
kmeans.fit(X)

# Predict the cluster for a new customer
new_customer = np.array([[55000, 1000]])  # Income: 55K, Spending: 1000
cluster = kmeans.predict(new_customer)
print(f"New customer belongs to Cluster: {cluster[0]}")
```

#  Learning Through Rewards & Penalties (Reinforcement Learning)

```python
import random

# The environment: A hidden number (agent must guess)
hidden_number = random.randint(1, 10)
attempts = 0
reward = 0

while True:
    guess = random.randint(1, 10)  # Agent randomly guesses a number
    attempts += 1
    
    if guess == hidden_number:
        reward += 10  # Reward for correct guess
        print(f"Guessed correctly in {attempts} attempts! Total reward: {reward}")
        break
    else:
        reward -= 1  # Penalty for wrong guess
        print(f"Wrong guess ({guess}). Penalty applied. Total reward: {reward}")
```

```python

from sklearn.linear_model import LinearRegression
import numpy as np

# Input (Square footage) and Output (House price)
X = np.array([[1000], [1500], [2000], [2500], [3000]])  # Features
y = np.array([200000, 250000, 300000, 350000, 400000])  # Labels

# Train a Linear Regression model
model = LinearRegression()
model.fit(X, y)

# Predict price for a 2200 sq ft house
predicted_price = model.predict([[2200]])
print(f"Predicted price for 2200 sq ft house: ${predicted_price[0]:,.2f}")
```

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB

# Training data (emails) and their labels (spam=1, not spam=0)
emails = ["Win a free iPhone", "Important meeting tomorrow", "Claim your lottery prize", "Team project discussion"]
labels = [1, 0, 1, 0]  # 1 = Spam, 0 = Not Spam

# Convert text data into numerical format
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(emails)

# Train a Naive Bayes classifier
model = MultinomialNB()
model.fit(X, labels)

# Predict if a new email is spam
new_email = ["Congratulations! You've won a free vacation"]
X_new = vectorizer.transform(new_email)
prediction = model.predict(X_new)

print(f"Prediction: {'Spam' if prediction[0] == 1 else 'Not Spam'}")
```
