# Random Forest classifier Alg:
Random forest is an ensemble learning technique.

Random forest builds multiple decision Trees and mergs them together to get a mpre accurate and stable prediction.

Then it gets predictions from each tree and by means of majority voting, it selects the decision which gets the majority vote.

## steps:
##### step 1:pick at random k data points from the training set.
##### step 2:Build the decision Tree associated with these k data points
##### step 3:Choose the number N tree of trees you want to build and repeat STEP1 & 2.
##### step 4:for a new data point, make each one of your Ntree Trees predict the category to which the data points belongs, and assign the new data point to the category that wins the majority vote.

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as pe
```

```python
# load the data set
df=pd.read_csv(r"C:\Users\Windows\Downloads\Social_Network_Ads (1).csv")
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
X=df.iloc[:,2:4].values
Y=df.iloc[:,-1].values
```

```python
# splittingg the data
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test=train_test_split(X, Y, random_state=0,test_size=0.25)
```

```python
from sklearn.ensemble import RandomForestClassifier
clf=RandomForestClassifier(n_estimators=10, criterion='entropy')
clf.fit(x_train, y_train)
```

```python
# predicting the test set results
y_pred=clf.predict(x_test)
```

```python
from sklearn.metrics import confusion_matrix
cm=confusion_matrix(y_test, y_pred)
sns.heatmap(cm, annot=True)
```
