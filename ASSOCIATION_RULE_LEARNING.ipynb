# Association Rule Learning

Association Rule Learning is a machine learning method used to discover intersting relationships (associations) between variables in large datasets. It's commonly used in market nasket analysis to find patterns llike , "If a customer buys bread, they are likely to buy butter."

# 🚀 1. Apriori Algorithm

works by identifying the frequent individual items and extending them to larger itemsets as long as they the minimum aupport threshold.

# STEPS:
1. identify frequently individual items (single-items sets).
2. Extend to lartger itemsets (like {bread, butter}) if they meet the minimum support.
3. Generate association rules form these itemsets usinig confidence and fit.

```python
# Example.
import warnings 
warnings.filterwarnings('ignore')
from mlxtend.frequent_patterns import apriori, association_rules
import pandas as pd

# Sample transaction data
data = {'bread': [1, 1, 0, 1, 0],
        'butter': [1, 1, 0, 0, 0],
        'milk': [0, 1, 1, 1, 1],
        'eggs': [0, 0, 1, 1, 1]}
df = pd.DataFrame(data)

# Apply Apriori algorithm
frequent_itemsets = apriori(df, min_support=0.5, use_colnames=True)
# frequent_itemsets
# Generate association rules
rules = association_rules(frequent_itemsets, metric='confidence', min_threshold=0.7)
# print(rules)
print(rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']])
```

# ⚡ 2. Eclat Algorithm
Focuses on itemset intersections using a depth-first search approach.
It's faster than Apriori for datasets with fewer transactions but more items.

# ⚡ 3. FP-Growth Algorithm

Uses a Frequent Pattern Tree (FP-Tree) to encode data, reducing the number of scans required.
More efficient than Apriori for large datasets.

```python
from mlxtend.frequent_patterns import fpgrowth

# Apply FP-Growth algorithm
frequent_itemsets = fpgrowth(df, min_support=0.5, use_colnames=True)

# Generate association rules
rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.7)
print(rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']])
```
