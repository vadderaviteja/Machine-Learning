```python
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
df=pd.read_csv(r"C:\Users\Windows\Downloads\Social_Network_Ads.csv")
```

```python

df.head()
```

```python
df.isna().sum()
```

```python
X=df.iloc[:, 2:4].values
Y=df.iloc[:,-1].values
```

```python
sns.countplot(data=df,x='Purchased')
```

# Undersampling

```python
from imblearn.under_sampling import NearMiss
nm=NearMiss()
x_res, y_res =nm.fit_resample(X,Y)
```

```python
X.shape
```

```python
x_res.shape
```

```python
from collections import Counter
print(f'Before under sampling :{Counter(Y)}')
print(f'Before under sampling :{Counter(y_res)}')
```

# Over Sampling

```python
from imblearn.combine import SMOTETomek
sm=SMOTETomek()
X_res, Y_res=sm.fit_resample(X, Y)
```

```python
print(f'Before over sampling :{Counter(Y)}')
print(f'Before over sampling :{Counter(Y_res)}')
```

```python
X.shape
```

```python
X_res.shape
```

# Random over sampling

```python
from imblearn.over_sampling import RandomOverSampler
ros=RandoOverSampler()
X_res, Y_res=ros.fit_resample(X,Y)
```

```python
print(f'Before RandomOverSampler :{Counter(Y)}')
print(f'Before RandomOverSampler :{Counter(Y_res)}')
```
